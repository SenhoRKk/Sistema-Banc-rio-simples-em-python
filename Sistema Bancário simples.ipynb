```python
def adicionar_receita(saldo, valor):
  return saldo + valor

def adicionar_despesa(saldo, valor):
  return saldo - valor

def consultar_saldo(saldo):
  return saldo

def mostrar_movimentacoes(movimentacoes):
  for numero, movimentacao in enumerate(movimentacoes, start=1):
    print(f"{numero} - {movimentacao}")
def consultar_totais(total_receitas, total_despesas):
  print(f"Total de receitas:R$ {total_receitas:.2f}")
  print(f"Total de despesas:R$ {total_despesas:.2f}")

saldo = float(input("Digite seu saldo: "))
movimentacoes = []
total_receitas = 0
total_despesas = 0


while True:
   operacao = int(input("Escolha uma opção: 1-Adicionar receita, 2-Adicionar despesas, 3-Consultar saldo, 4-Mostrar movimentações, 5- Consultar totais, 6-sair "))
   if operacao == 6:
    print("Até mais!")
    break
   elif operacao == 1:
    valor = float(input("Digite o valor da receita: "))
    if valor < 0:
      print("Valor inválido!")
    else:
       total_receitas = total_receitas + valor
       saldo = adicionar_receita(saldo, valor)
       movimentacoes.append(f"Receita:  $ {valor:.2f}")
       print("Receita adicionada com sucesso!")

   elif operacao == 2:
    valor = float(input("Digite o valor da despesa: "))
    if valor < 0:
      print("Valor inválido!")
    elif valor > saldo:
      print("Saldo insuficiente!")
    else:
       saldo = adicionar_despesa(saldo, valor)
       total_despesas = total_despesas + valor
       movimentacoes.append(f"Despesa:  $ {valor}")
       print("Despesa adicionada com sucesso!")
   elif operacao == 3:
    print("Seu saldo é: ", consultar_saldo(saldo))

   elif operacao == 4:
    mostrar_movimentacoes(movimentacoes)
   elif operacao == 5:
    consultar_totais(total_receitas, total_despesas)
   else:
    print("Opção inválida!")

```
